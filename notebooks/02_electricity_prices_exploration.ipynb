### There is only one missing in 2007. However, I want to keep year 2007.

### Creating a copy of the dataframe, and continuing the transformation.

```python
electricity_df = filtered_df_detailed.copy()
```

```python
electricity_df.head(2)
```

### Changing wide format into long format. For that purpose melt function will be used.

```python
# Select period columns (2007-2024) and exclude incomplete 2025 data
period_columns = []

for col in electricity_df.columns:
    if col.startswith("2") and not col.startswith("2025"):
        period_columns.append(col)

print(period_columns)
```

```python
electricity_melted_df = electricity_df.melt(
    id_vars=['geo\\TIME_PERIOD'],
    value_vars = period_columns,
    var_name='period',
    value_name = 'electricity_price'
)

```

```python
electricity_melted_df.shape
```

```python
#final_electricity_df.to_csv("../data/processed/electricity_prices_poland_eu.csv",
                            #index = False
                           #)
```

```python
electricity_melted_df.head(10)
```

### After melt function there are 72 rows (2 countries with 36 rows with period 2007-S1 until 2024-S2 were melted into 72 rows as each period becomes separated row). There are also 3 columns: 'geo\TIME_PERIOD,'period', and 'electricity_price'. Changing name of first column to 'country', as this is more readible.

```python
electricity_melted_df.rename(columns={'geo\\TIME_PERIOD':
'country'},inplace=True)
```

```python
electricity_melted_df.head(10)
```

### Separating number from semester in period column.

```python
electricity_melted_df["year"] = electricity_melted_df["period"].str.split("-").str[0]
electricity_melted_df["semester"] = electricity_melted_df["period"].str.split("-").str[1]

print(electricity_melted_df)
```

### Removing period column, as it is no longer needed.

```python
electricity_melted_df.drop("period",axis = 1, inplace = True)
print(electricity_melted_df)
```

```python
electricity_melted_df.isna().sum()
```

```python
electricity_melted_df.dtypes
```

### Grouping the data.

```python
electricity_grouped_df = electricity_melted_df.groupby(["country", "year"])[["electricity_price"]].mean()
```

### Reseting MultiIndex back to "normal index" with normal columns and default digital index and limiting decimals to 4 places.

```python
electricity_grouped_df.reset_index()
```

```python
electricity_grouped_df = electricity_grouped_df["electricity_price"].round(4)
```
