Link: https://ourworldindata.org/co2-and-greenhouse-gas-emissions

## Load dataset

```python
import pandas as pd

#Fetch the data
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)
```

## Saving raw dataset

```python
df.to_csv(
    "../data/raw/owid_co2_data.csv",
    index=False
)
```

## The purpose of this section is to understand the dataset structure, verify data quality, inspect data types, identify missing values, and determine which variables are relevant for further analysis.

### Checking columns in the dataset. 

```python
df.columns.tolist()
```

### Selecting 5 columns for further analysis: population, country, year, co2, and co2_per_capita for my analysis.

```python
df.shape
```

### Let's see the structure

```python
df.head(10)
```

### In data there are missing values. Veifying whether they are also in the selected columns

```python
selected_columns = [
    "country",
    "year",
    "population",
    "co2",
    "co2_per_capita"
]

df[selected_columns].isna().sum()
```

### Some missing values are present in the selected columns.
### The next step is to verify whether they affect Poland or the EU27 observation

```python
min_year = df["year"].min()
max_year = df["year"].max()

print(f"Minimum year is {min_year}, and maximum is {max_year}")
```

### The latest data is 2024, and earliest 1750. Checking data types

```python
df.info()
```

### Checking unique country values to see whether we look for "EU", or "European Union" etc.

```python
sorted(df["country"].unique())
```

### Data includes European Union with 27, or 28 countries. EU 27 countries will be analyzed (without Great Britain).

```python
country_filter = df["country"].isin(["Poland", "European Union (27)"])
year_filter = df["year"] >= 1990

filtered_df = df[country_filter & year_filter]
```

### Filtering data for desired columns: population, year, country, co2, and co2_per_capita

```python
filtered_df = filtered_df[selected_columns]
```

### Check to see whether there are still some missing values

```python
filtered_df.isna().sum()
```

```python
filtered_df.info()
```

### The population column is stored as float64 because the original dataset contains missing values for some countries and years. The filtered dataset contains no missing values, so the column could be safely converted to int64 if required.

```python
filtered_df.head(10)
```

### There could be a possibility to convert 64 bit data types into something smaller to save space if required. I also noticed old index is kept. Resseting the old index.


```python
filtered_df = filtered_df.reset_index(drop=True)
```

## Data quality summary

- Period: 1990–2024
